```python
from sklearn.datasets import fetch_20newsgroups
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB, BernoulliNB
from sklearn.metrics import accuracy_score, f1_score, classification_report
print("Loading dataset...")

train_data = fetch_20newsgroups(
    subset='train',
    remove=('headers', 'footers', 'quotes')
)

test_data = fetch_20newsgroups(
    subset='test',
    remove=('headers', 'footers', 'quotes')
)

X_train = train_data.data
y_train = train_data.target

X_test = test_data.data
y_test = test_data.target

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
print("Number of categories:", len(train_data.target_names))

vectorizer = TfidfVectorizer(
    stop_words='english',
    max_features=20000
)

X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)

print("\nNumber of features:", X_train_tfidf.shape[1])

multinomial_nb = MultinomialNB()

multinomial_nb.fit(X_train_tfidf, y_train)

y_pred_multinomial = multinomial_nb.predict(X_test_tfidf)

multinomial_accuracy = accuracy_score(
    y_test,
    y_pred_multinomial
)

multinomial_f1 = f1_score(
    y_test,
    y_pred_multinomial,
    average='weighted'
)

print("\n----- Multinomial Naive Bayes -----")
print("Accuracy:", round(multinomial_accuracy, 4))
print("Weighted F1-score:", round(multinomial_f1, 4))


bernoulli_nb = BernoulliNB(
    binarize=0.0
)

bernoulli_nb.fit(X_train_tfidf, y_train)

y_pred_bernoulli = bernoulli_nb.predict(X_test_tfidf)

bernoulli_accuracy = accuracy_score(
    y_test,
    y_pred_bernoulli
)

bernoulli_f1 = f1_score(
    y_test,
    y_pred_bernoulli,
    average='weighted'
)

print("\n----- Bernoulli Naive Bayes -----")
print("Accuracy:", round(bernoulli_accuracy, 4))
print("Weighted F1-score:", round(bernoulli_f1, 4))


print("\n========== Model Comparison ==========")

print("Multinomial NB")
print("Accuracy:", round(multinomial_accuracy, 4))
print("Weighted F1-score:", round(multinomial_f1, 4))

print("\nBernoulli NB")
print("Accuracy:", round(bernoulli_accuracy, 4))
print("Weighted F1-score:", round(bernoulli_f1, 4))

print("\n===== Classification Report: Multinomial NB =====")

print(classification_report(
    y_test,
    y_pred_multinomial,
    target_names=test_data.target_names
))
```
